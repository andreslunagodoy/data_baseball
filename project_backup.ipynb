## Creating Database

```python
#Create the SQLite Database
import sqlite3

conn = sqlite3.connect(DB_FILE)
cursor = conn.cursor()
```

```python
#Load all .csv files into SQLite
for file in csv_files:
    table_name = file.replace(".csv", "")
    file_path = os.path.join(DATA_DIR, file)
    
    print(f"Loading {table_name}...")
    
    df = pd.read_csv(file_path)
    
    df.to_sql(
        table_name,
        conn,
        if_exists="replace",
        index=False
    )

print("All tables loaded.")
```

```python
#Verify tables exist
query = """
SELECT name 
FROM sqlite_master 
WHERE type='table'
ORDER BY name;
"""

pd.read_sql(query, conn)
```

```python
#Inspect a table schema
#Example: Batting
pd.read_sql("PRAGMA table_info(Batting);", conn)
```

```python
#Sanity check
#Check that playerID overlaps
pd.read_sql("""
SELECT COUNT(DISTINCT playerID) 
FROM Batting;
""", conn)
```

```python
pd.read_sql("""
SELECT COUNT(DISTINCT playerID) 
FROM People;
""", conn)
```

```python
pd.read_sql("""
SELECT p.nameFirst, p.nameLast, b.yearID, b.HR
FROM Batting b
JOIN People p ON b.playerID = p.playerID
LIMIT 10;
""", conn)
```

```python
#Create most important indices
cursor.execute("CREATE INDEX idx_batting_playerID ON Batting(playerID);")
cursor.execute("CREATE INDEX idx_batting_yearID ON Batting(yearID);")
cursor.execute("CREATE INDEX idx_pitching_playerID ON Pitching(playerID);")
cursor.execute("CREATE INDEX idx_people_playerID ON People(playerID);")

conn.commit()
```
